## 2.1. Variables y tipos de dato

Una **variable** es una caja con nombre donde guardas un valor. Python reconoce solo el tipo que le pongas:

| Tipo | Nombre en Python | Ejemplo | Cuándo lo usamos aquí |
|---|---|---|---|
| Texto | `str` | `"Corea, República de"` | Nombres de país, códigos con ceros a la izquierda |
| Número entero | `int` | `2024` | Años |
| Número decimal | `float` | `19316.0` | Valores monetarios, índices |
| Verdadero/falso | `bool` | `True` | Filtros y condiciones |

Esta distinción **no es un detalle técnico**: más adelante verás que si Python lee el código de país `"040"` (Austria) como número, lo convierte en `40` y pierde el cero inicial. Con eso se rompe el cruce con la otra base de datos y el análisis queda mal. Un error de tipo de dato es un error de negocio.

```python
# Cuatro variables, una de cada tipo
pais_candidato = "Corea, Republica de"          # str  (texto)
anio_referencia = 2024                           # int  (entero)
importaciones_kusd = 19316.0                     # float (decimal)
es_socio_actual = False                          # bool (verdadero/falso)

# type() nos dice de que tipo es cada cosa
print(pais_candidato, "->", type(pais_candidato))
print(anio_referencia, "->", type(anio_referencia))
print(importaciones_kusd, "->", type(importaciones_kusd))
print(es_socio_actual, "->", type(es_socio_actual))
```

```python
# El problema de los ceros a la izquierda, en vivo:
codigo_como_texto = "040"     # asi viene en el archivo de Trade Map (Austria)
codigo_como_numero = int("040")

print("Como texto :", codigo_como_texto)
print("Como numero:", codigo_como_numero)
print("Son iguales?", codigo_como_texto == str(codigo_como_numero))
```

El código quedó convertido de `"040"` a `40`. Cuando más adelante intentemos emparejar ese código con otra tabla que sí conserva el `"040"`, el emparejamiento fallará en silencio: no aparecerá ningún error, simplemente ese país se quedará sin datos. **Los errores silenciosos son los peligrosos.**

## 2.3. Funciones

Una **función** es una receta con nombre. La escribes una vez y la usas muchas veces. Si mañana cambia la receta, la corriges en un solo lugar.

En este cuaderno vamos a escribir dos funciones importantes: una que limpia cualquier archivo de Trade Map y otra que limpia cualquier archivo del Banco Mundial. Gracias a ellas, limpiar cinco archivos costará casi lo mismo que limpiar uno.

Empecemos con una función pequeña y muy pertinente: Trade Map entrega los valores en **miles de dólares** (kUSD), pero las fórmulas del documento base trabajan en **dólares**.

```python
def kusd_a_usd(valor_en_miles):
    '''Convierte un valor expresado en miles de dolares (kUSD) a dolares (USD).

    Trade Map reporta 'Value (kUSD)'. Multiplicar por 1000 evita que despues
    dividamos dolares entre miles de dolares y obtengamos un indice absurdo.
    '''
    return valor_en_miles * 1000


# Probemos la funcion con el dato real de Corea del Sur
importaciones_corea_usd = kusd_a_usd(19316.0)
print(f"Corea del Sur importo USD {importaciones_corea_usd:,.0f} en claveles frescos.")
```

**Para pensar.** El error de mezclar unidades no es teórico. Si calculas el Índice de Apertura Comercial dividiendo un comercio expresado en miles de dólares entre un PIB expresado en dólares, el resultado sale mil veces más pequeño y podrías concluir que una economía muy abierta es cerrada. Ese es el tipo de equivocación que llega a un comité de inversión.

# 3. pandas: la hoja de cálculo con superpoderes

## 3.1. Qué es

**pandas** es la biblioteca de Python que se usa para trabajar con datos en forma de tabla (McKinney, 2010; The pandas development team, 2025). Piensa en Excel, pero:

- sin límite práctico de filas,
- donde cada paso queda escrito y por lo tanto es **auditable y repetible**,
- y donde repetir el análisis el próximo trimestre cuesta un clic, no un día.

Esa capacidad de auditar es exactamente el principio metodológico que exige Baena-Rojas y Cano (2026): cada variable debe tener una fuente pública, replicable y fechada, "evitando así estimaciones discrecionales que un gerente no pueda auditar ni un competidor pueda refutar con datos" (Serie de recursos, 2026, p. 7).

## 3.2. Las dos estructuras de pandas

| Estructura | Qué es | Analogía en Excel |
|---|---|---|
| `Series` | Una sola columna con etiquetas | Una columna |
| `DataFrame` | Una tabla completa: varias columnas con un índice común | Una hoja de cálculo |

Y una convención que vale para todo el curso: en una tabla bien construida,

- cada **fila** es una observación (aquí: un país),
- cada **columna** es una variable (aquí: importaciones, PIB, distancia),
- cada **celda** es un valor.

A esa forma se le llama **datos ordenados** (*tidy data*) y es el formato que necesitan todas las fórmulas del documento base.

## 4.2. Encontrar los archivos sin depender del nombre exacto

Los archivos del Banco Mundial traen un número de versión al final que cambia en cada descarga (`..._v2_234.csv`, `..._v2_35140.csv`). Si escribimos el nombre completo a mano, el cuaderno se rompe la próxima vez que alguien descargue los datos.

La solución es buscar por **patrón** con el módulo `glob`. El símbolo `*` significa "aquí puede ir cualquier cosa", y `**` con `recursive=True` significa "búscalo también dentro de las subcarpetas". Así funciona igual si subiste los archivos sueltos o si mantuviste la estructura de carpetas del Banco Mundial.

```python
def buscar_archivo(patron, ruta_base=None):
    '''Busca un archivo por patron dentro de la ruta base, incluyendo subcarpetas.

    Devuelve la ruta del archivo encontrado. Si hay varios, avisa y toma el primero.
    Si no hay ninguno, lanza un error claro en vez de fallar mas adelante sin explicacion.
    '''
    if ruta_base is None:
        ruta_base = RUTA_BASE

    coincidencias = sorted(glob.glob(os.path.join(ruta_base, "**", patron), recursive=True))

    if len(coincidencias) == 0:
        raise FileNotFoundError(
            f"No encontre ningun archivo que coincida con '{patron}' dentro de '{ruta_base}'.\n"
            f"Revisa que subiste el archivo y que el nombre no cambio."
        )
    if len(coincidencias) > 1:
        print(f"  Aviso: hay {len(coincidencias)} archivos que coinciden con '{patron}'. Uso el primero.")

    return coincidencias[0]


# Localizamos los cinco archivos que necesitamos
ARCHIVOS = {
    "tm_importaciones": buscar_archivo("importing-countries-*_060312.csv"),
    "tm_exportaciones": buscar_archivo("exporting-countries-*_060312.csv"),
    "bm_pib":           buscar_archivo("API_NY.GDP.MKTP.CD_DS2*.csv"),
    "bm_exportaciones": buscar_archivo("API_NE.EXP.GNFS.CD_DS2*.csv"),
    "bm_importaciones": buscar_archivo("API_NE.IMP.GNFS.CD_DS2*.csv"),
    "bm_metadatos":     buscar_archivo("Metadata_Country_API_NY.GDP.MKTP.CD*.csv"),
}

for nombre, ruta in ARCHIVOS.items():
    print(f"{nombre:18s} -> {os.path.basename(ruta)}")
```

## 5.1. Diagnóstico: nueve problemas en un solo archivo

Lo que acabas de ver parece una tabla decente. No lo es. Vamos por partes.

| # | Problema | Por qué importa |
|---|---|---|
| 1 | **Nombres de columna incómodos**: `Value (kUSD)`, `Growth Value 5Y (%)` | Tienen espacios, paréntesis y están en inglés. Complican el código y se prestan a errores de tipeo |
| 2 | **Códigos de país leídos como número** | `"040"` (Austria) se vuelve `40` y pierde el cero. Rompe el cruce con la otra base |
| 3 | **Una fila que no es un país**: `Mundo` (código `000`) | Es el total mundial. Si lo dejas, el "mayor importador del mundo" será el mundo entero |
| 4 | **Valores en miles de dólares**, no en dólares | Mezclar unidades con el PIB del Banco Mundial da índices absurdos |
| 5 | **Celdas vacías** en `Value (kUSD)` | Países que aparecen listados pero sin cifra reportada. ¿Son cero o son desconocidos? |
| 6 | **Unidades de cantidad mezcladas**: Toneladas, Unidad, Docena, No medida | El precio unitario de un país en USD/tonelada **no es comparable** con el de otro en USD/docena |
| 7 | **Columnas constantes**: `partnerCd`, `partnerLabel`, `productCd`, `productLabel` | Todas las filas dicen lo mismo. Ocupan espacio y no informan nada |
| 8 | **Precisión falsa**: valores unitarios con 28 decimales | Ninguna estadística de comercio tiene esa precisión. Da una falsa sensación de exactitud |
| 9 | **Ausencia de la columna de año** | El archivo se llama "2025" pero el año no está dentro de la tabla. Si lo mezclas con otro archivo, pierdes la trazabilidad |

## 6.10. Todo junto: una función reutilizable

Acabamos de dar nueve pasos sobre el archivo de importaciones. El archivo de exportaciones necesita exactamente los mismos nueve. Copiar y pegar el código sería un error: si mañana descubrimos que un paso estaba mal, tendríamos que acordarnos de corregirlo en dos lugares.

Escribimos entonces **una sola función** que recibe la ruta de cualquier archivo de Trade Map y devuelve la tabla limpia.

```python
def limpiar_trademap(ruta_archivo, etiqueta_flujo):
    '''Limpia un archivo CSV descargado de Trade Map (ITC).

    Parametros
    ----------
    ruta_archivo : str
        Ruta del CSV descargado de Trade Map.
    etiqueta_flujo : str
        'imp' o 'exp'. Se usa para nombrar las columnas del resultado.

    Devuelve
    --------
    tabla : DataFrame con una fila por pais.
    total_mundial : float, valor total mundial en USD (la fila 'Mundo', separada).
    '''
    # 1. Leer conservando los ceros a la izquierda de los codigos
    df = pd.read_csv(
        ruta_archivo,
        dtype={"reporterCd": str, "partnerCd": str, "productCd": str},
        encoding="utf-8",
    )

    # 2. Renombrar a nombres claros en espanol
    df = df.rename(columns=NOMBRES_TRADEMAP)

    # 3. Quitar columnas constantes
    df = df.drop(columns=["partnerCd", "partnerLabel", "productCd", "productLabel"])

    # 4. Convertir texto a numero (lo no convertible queda como NaN)
    for col in COLUMNAS_NUMERICAS:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    # 5. Pasar de miles de dolares a dolares y recortar la precision falsa
    df["valor_usd"] = kusd_a_usd(df["valor_kusd"])
    df["balanza_usd"] = kusd_a_usd(df["balanza_kusd"])
    df["valor_unitario"] = df["valor_unitario"].round(2)

    # 6. Separar el agregado mundial
    fila_mundo = df[df["codigo_tm"] == CODIGO_MUNDO]
    total_mundial = float(fila_mundo["valor_usd"].iloc[0]) if len(fila_mundo) else np.nan
    df = df[df["codigo_tm"] != CODIGO_MUNDO].copy()

    # 7. Conservar el valor unitario solo donde es comparable
    df["valor_unitario_usd_ton"] = df["valor_unitario"].where(df["unidad_cantidad"] == UNIDAD_COMPARABLE)
    df["cantidad_ton"] = df["cantidad"].where(df["unidad_cantidad"] == UNIDAD_COMPARABLE)

    # 8. Limpiar espacios sobrantes en los nombres de pais
    df["pais"] = df["pais"].str.strip()

    # 9. Renombrar las columnas del flujo para poder unir importaciones con exportaciones
    columnas_flujo = {
        "valor_usd": f"{etiqueta_flujo}_usd",
        "participacion_mundial_pct": f"{etiqueta_flujo}_participacion_mundial_pct",
        "crec_valor_5a_pct": f"{etiqueta_flujo}_crec_5a_pct",
        "crec_valor_2a_pct": f"{etiqueta_flujo}_crec_2a_pct",
        "crec_cantidad_5a_pct": f"{etiqueta_flujo}_crec_cantidad_5a_pct",
        "valor_unitario_usd_ton": f"{etiqueta_flujo}_valor_unitario_usd_ton",
        "cantidad_ton": f"{etiqueta_flujo}_cantidad_ton",
        "unidad_cantidad": f"{etiqueta_flujo}_unidad_cantidad",
        "distancia_km": f"{etiqueta_flujo}_distancia_km",
        "concentracion": f"{etiqueta_flujo}_concentracion",
        "balanza_usd": f"{etiqueta_flujo}_balanza_reportada_usd",
    }
    df = df.rename(columns=columnas_flujo)

    columnas_finales = ["codigo_tm", "pais"] + list(columnas_flujo.values())
    return df[columnas_finales].reset_index(drop=True), total_mundial


print("Funcion limpiar_trademap() definida.")
```

```python
# Una funcion, dos archivos
tm_imp, mundo_imp_usd = limpiar_trademap(ARCHIVOS["tm_importaciones"], "imp")
tm_exp, mundo_exp_usd = limpiar_trademap(ARCHIVOS["tm_exportaciones"], "exp")

print(f"Importaciones: {tm_imp.shape[0]} paises, {tm_imp.shape[1]} columnas")
print(f"Exportaciones: {tm_exp.shape[0]} paises, {tm_exp.shape[1]} columnas")
print(f"\nTotal mundial importado : USD {mundo_imp_usd:,.0f}")
print(f"Total mundial exportado : USD {mundo_exp_usd:,.0f}")
print(f"Diferencia               : USD {mundo_imp_usd - mundo_exp_usd:,.0f}")
```

**Para pensar.** El total mundial importado y el total mundial exportado deberían ser iguales: lo que un país exporta, otro lo importa. No lo son, y la diferencia es de unos 101 millones de dólares.

Esto no es un error del ejercicio, es un fenómeno bien conocido de las estadísticas de comercio: las importaciones se suelen reportar en valor CIF (incluyen flete y seguro) y las exportaciones en valor FOB (no los incluyen), hay diferencias de fecha de registro entre el embarque y la llegada, y no todos los países reportan. Anotarlo es parte del oficio; ignorarlo es lo que produce análisis frágiles.

## 6.11. Unir importaciones y exportaciones

Ahora juntamos las dos tablas para tener, en una sola fila, cuánto importó y cuánto exportó cada país. Esa fila es exactamente lo que necesita la fórmula del IBCR del Cuaderno 2.

El tipo de unión importa:

| Tipo | Qué conserva |
|---|---|
| `inner` | Solo los países que están en **ambas** tablas |
| `outer` | **Todos** los países de las dos tablas |
| `left` | Todos los de la tabla izquierda |

Usamos `outer` porque no queremos perder ningún país por el camino, y porque queremos ver explícitamente quién aparece en una tabla y no en la otra.

```python
comercio = tm_imp.merge(
    tm_exp.drop(columns=["pais"]),      # el nombre del pais ya viene en tm_imp
    on="codigo_tm",
    how="outer",
    indicator=True,                      # anade una columna que dice de donde salio cada fila
)

print("Origen de cada fila tras la union:\n")
print(comercio["_merge"].value_counts().to_string())
print(f"\nTotal de paises: {len(comercio)}")

comercio = comercio.drop(columns=["_merge"])
```

## 6.12. Validar la decisión sobre los faltantes

Llegó el momento de resolver la duda del paso 6.7 con evidencia en vez de con suposiciones.

Trade Map trae una columna `Balance (kUSD)` que reporta la balanza comercial del producto. Por definición:

$$\text{Balanza} = X - M$$

Si nosotros reconstruimos la balanza con nuestras propias columnas de exportaciones e importaciones y el resultado coincide con la balanza que reporta Trade Map, entonces nuestra lectura de los datos es correcta. Si además coincide **tratando los faltantes como cero**, tenemos evidencia de que esos faltantes efectivamente significan "no hubo comercio registrado" y no "no sabemos".

Esta es una **prueba de consistencia interna**, y es una de las herramientas más valiosas del oficio: usar una columna que no utilizaste para verificar el trabajo que hiciste con las otras.

```python
# Tratamos los faltantes como cero SOLO para la prueba, en columnas auxiliares
prueba_m = comercio["imp_usd"].fillna(0)
prueba_x = comercio["exp_usd"].fillna(0)

balanza_reconstruida = prueba_x - prueba_m
balanza_reportada = comercio["imp_balanza_reportada_usd"].fillna(comercio["exp_balanza_reportada_usd"])

diferencia = (balanza_reconstruida - balanza_reportada).abs()

print(f"Paises comparados                : {diferencia.notna().sum()}")
print(f"Diferencia maxima encontrada     : USD {diferencia.max():,.0f}")
print(f"Paises con diferencia mayor a $1 : {(diferencia > 1).sum()}")
```

La diferencia es cero. Nuestra reconstrucción coincide exactamente con la balanza que reporta la fuente.

**Conclusión con evidencia:** en este archivo, una celda vacía en el valor de importación o exportación significa que el país no registró comercio de claveles frescos en 2025. Ahora sí podemos reemplazar esos faltantes por cero, **y podemos justificarlo por escrito ante quien nos pregunte**. Esa es la diferencia entre limpiar datos y manipularlos.

```python
comercio["imp_usd"] = comercio["imp_usd"].fillna(0)
comercio["exp_usd"] = comercio["exp_usd"].fillna(0)

# Columnas derivadas que usaremos en el Cuaderno 2
comercio["comercio_total_usd"] = comercio["imp_usd"] + comercio["exp_usd"]
comercio["balanza_usd"] = comercio["exp_usd"] - comercio["imp_usd"]

# Recalculamos la participacion mundial con nuestro propio denominador, para que sea trazable
comercio["imp_participacion_mundial_pct"] = comercio["imp_usd"] / mundo_imp_usd * 100
comercio["exp_participacion_mundial_pct"] = comercio["exp_usd"] / mundo_exp_usd * 100

print("Los diez mayores importadores mundiales de claveles frescos (HS 060312), 2025:\n")
top10 = comercio.nlargest(10, "imp_usd")[
    ["pais", "imp_usd", "imp_participacion_mundial_pct", "imp_crec_5a_pct", "imp_valor_unitario_usd_ton"]
]
print(top10.to_string(index=False))
```

Este es el primer resultado con valor de negocio del cuaderno: **el ranking real de compradores mundiales de claveles**, ya sin el agregado "Mundo" contaminándolo y con los valores en dólares.

Nota dos cosas que van a ser centrales en el Cuaderno 2:

- **Corea del Sur no está en el top 5.** Está más abajo. El caso de negocio, entonces, no es "vamos al mercado más grande", sino algo más interesante: ¿hay razones para preferir un mercado mediano sobre uno grande?
- **Países Bajos aparece segundo**, pero también es uno de los mayores exportadores del mundo. Eso debería levantar una sospecha que resolveremos con el IBCR.

| 4 | **Filas que no son países** | "Africa Eastern and Southern", "Euro area", "High income": son agregados regionales y grupos de ingreso |
| 5 | **Años sin dato** | Los años más recientes están vacíos para muchos países |
| 6 | **Otro idioma de códigos** | Aquí los países se identifican con códigos ISO de tres letras (`KOR`, `COL`), no con números como en Trade Map |

Los resolvemos en orden.

## 7.2. Saltar las líneas de metadatos

`skiprows=4` le dice a pandas: *"ignora las primeras cuatro líneas; el encabezado real está en la quinta"*.

```python
bm = pd.read_csv(ARCHIVOS["bm_pib"], skiprows=4)

print("Forma:", bm.shape)
print("Primeras columnas:", list(bm.columns[:6]))
print("Ultimas columnas :", list(bm.columns[-4:]))
bm.head(3)
```

## 7.3. De formato ancho a formato largo

Este es el paso conceptualmente más importante de la sección, así que vale la pena entenderlo bien.

**Formato ancho** (como viene el archivo): cada año es una columna.

| País | 2022 | 2023 | 2024 |
|---|---|---|---|
| Corea | 1.674 | 1.845 | 1.875 |

**Formato largo** (lo que queremos): cada combinación de país y año es una fila.

| País | año | PIB |
|---|---|---|
| Corea | 2022 | 1.674 |
| Corea | 2023 | 1.845 |
| Corea | 2024 | 1.875 |

¿Por qué el formato largo es mejor? Porque en él, **el año es un dato**, no un nombre de columna. Y como es un dato, puedes filtrarlo, agruparlo, compararlo y unirlo con otras tablas usando las herramientas normales de pandas. En formato ancho, cambiar de año implica cambiar el código; en formato largo, implica cambiar un valor.

Esta es la forma de **datos ordenados** (*tidy data*) que mencionamos en la sección 3, y la operación que hace la transformación se llama `melt` ("derretir").

```python
# Identificamos las columnas de anio: son las que tienen un nombre formado solo por digitos.
# Esto tambien resuelve el PROBLEMA 3 (la columna fantasma queda fuera automaticamente).
columnas_anio = [c for c in bm.columns if c.isdigit()]

print(f"Detecte {len(columnas_anio)} columnas de anio, de {columnas_anio[0]} a {columnas_anio[-1]}.")
print("Columnas que quedan fuera:", [c for c in bm.columns if not c.isdigit()])
```

```python
bm_largo = bm.melt(
    id_vars=["Country Name", "Country Code"],   # columnas que se quedan quietas
    value_vars=columnas_anio,                   # columnas que se convierten en filas
    var_name="anio",                            # nombre de la nueva columna de etiquetas
    value_name="pib_usd",                       # nombre de la nueva columna de valores
)

bm_largo["anio"] = bm_largo["anio"].astype(int)

print("Antes (ancho):", bm.shape)
print("Despues (largo):", bm_largo.shape)
print()
print(bm_largo[bm_largo["Country Code"] == "KOR"].tail(5).to_string(index=False))
```

## 7.5. Separar los países de los agregados

Este es el mismo problema conceptual de la fila "Mundo" en Trade Map, pero disfrazado y multiplicado. El archivo del Banco Mundial mezcla 217 países reales con 47 agregados: regiones ("Latin America & Caribbean"), grupos de ingreso ("High income"), bloques ("Euro area") y el total mundial.

Si no los separas y calculas, por ejemplo, "el país con mayor PIB", el resultado será "Mundo". Y si calculas un promedio, estarás promediando países con continentes enteros.

La buena noticia es que el Banco Mundial nos da la herramienta para distinguirlos: el archivo `Metadata_Country`. En él, **los agregados son justamente las filas que no tienen región asignada**, porque un agregado no pertenece a ninguna región: la contiene.

```python
meta = pd.read_csv(ARCHIVOS["bm_metadatos"])

print("Columnas del archivo de metadatos:", list(meta.columns[:5]))
print(f"\nFilas totales      : {len(meta)}")
print(f"Con region asignada: {meta['Region'].notna().sum()}  <- paises reales")
print(f"Sin region         : {meta['Region'].isna().sum()}  <- agregados")
print("\nEjemplos de agregados:")
print(meta[meta["Region"].isna()]["Country Code"].head(12).tolist())
```

## 7.6. Una función para los tres indicadores

Los tres archivos del Banco Mundial (PIB, exportaciones totales, importaciones totales) tienen exactamente la misma estructura. Aplicamos la misma lógica de antes: una función, tres usos.

```python
def limpiar_banco_mundial(ruta_archivo, nombre_columna, codigos_validos):
    '''Limpia un archivo CSV de indicadores del Banco Mundial (formato World Development Indicators).

    Parametros
    ----------
    ruta_archivo : str
        Ruta del CSV descargado del Banco Mundial.
    nombre_columna : str
        Nombre que tendra la columna de valores en el resultado.
    codigos_validos : set
        Codigos ISO alfa-3 que corresponden a paises reales (no agregados).

    Devuelve
    --------
    DataFrame en formato largo con columnas: iso3, pais_bm, anio, <nombre_columna>.
    '''
    # 1. Saltar las cuatro lineas de metadatos
    df = pd.read_csv(ruta_archivo, skiprows=4)

    # 2. Detectar las columnas de anio (y descartar la columna fantasma)
    cols_anio = [c for c in df.columns if c.isdigit()]

    # 3. Pasar de formato ancho a formato largo
    df = df.melt(
        id_vars=["Country Name", "Country Code"],
        value_vars=cols_anio,
        var_name="anio",
        value_name=nombre_columna,
    )
    df["anio"] = df["anio"].astype(int)

    # 4. Quitar los anios sin dato
    df = df.dropna(subset=[nombre_columna])

    # 5. Conservar solo paises reales
    df = df[df["Country Code"].isin(codigos_validos)]

    # 6. Nombres de columna claros
    df = df.rename(columns={"Country Code": "iso3", "Country Name": "pais_bm"})

    return df[["iso3", "pais_bm", "anio", nombre_columna]].reset_index(drop=True)


print("Funcion limpiar_banco_mundial() definida.")
```

```python
pib   = limpiar_banco_mundial(ARCHIVOS["bm_pib"],           "pib_usd",           codigos_pais)
xtot  = limpiar_banco_mundial(ARCHIVOS["bm_exportaciones"], "exportaciones_totales_usd", codigos_pais)
mtot  = limpiar_banco_mundial(ARCHIVOS["bm_importaciones"], "importaciones_totales_usd", codigos_pais)

print(f"PIB                    : {len(pib):,} filas")
print(f"Exportaciones totales  : {len(xtot):,} filas")
print(f"Importaciones totales  : {len(mtot):,} filas")
```

```python
# Unimos los tres indicadores por pais y anio.
# Usamos 'inner': solo conservamos las combinaciones pais-anio que tienen los TRES datos,
# porque el Indice de Apertura necesita los tres a la vez.
macro = (
    pib.merge(xtot, on=["iso3", "pais_bm", "anio"], how="inner")
       .merge(mtot, on=["iso3", "pais_bm", "anio"], how="inner")
)

print(f"Combinaciones pais-anio completas: {len(macro):,}")
print(f"Paises                            : {macro['iso3'].nunique()}")
print(f"Rango de anios                    : {macro['anio'].min()} a {macro['anio'].max()}")
```

## 7.7. Elegir un año de referencia comparable

Aquí hay una trampa que atrapa a mucha gente. Es tentador tomar, para cada país, el **último año disponible**. Parece razonable: usar el dato más fresco de cada uno.

Es un error metodológico. Si comparas el PIB de Estados Unidos de 2024 con el de Polonia de 2025, no estás comparando dos países: estás comparando dos países **y** dos momentos económicos distintos. La conclusión mezcla las dos cosas y no puedes separarlas.

La regla es: **para comparar países entre sí, todos deben estar en el mismo año.** Y ese año es el más reciente en el que *todos* los países que te interesan tienen dato completo.

Veamos cuál es.

```python
# ¿Cuantos paises tienen dato completo en cada uno de los ultimos anios?
print("Cobertura por anio (paises con PIB, exportaciones e importaciones):\n")
print(macro["anio"].value_counts().sort_index().tail(6).to_string())
```

```python
ANIO_REFERENCIA = 2024

macro_ref = macro[macro["anio"] == ANIO_REFERENCIA].copy()

print(f"Anio de referencia elegido: {ANIO_REFERENCIA}")
print(f"Paises con datos completos: {len(macro_ref)}")
print()
print(macro_ref.head(5).to_string(index=False))
```

**Nota de trazabilidad.** Los datos de comercio de claveles son de 2025 y los datos macroeconómicos son de 2024. No es lo ideal, pero es lo que hay disponible, y **decirlo explícitamente es obligatorio**. El desfase de un año no invalida el análisis porque el PIB y la apertura comercial de un país son variables estructurales que se mueven lentamente. Lo que sí sería inaceptable es no advertirlo.

# 8. Unir dos mundos: el problema de las llaves

Tenemos dos tablas limpias que hablan del mismo planeta, pero no se entienden entre ellas:

| | Identifica a Corea del Sur como | Identifica a Estados Unidos como |
|---|---|---|
| **Trade Map** | `"410"` | `"842"` |
| **Banco Mundial** | `"KOR"` | `"USA"` |

Para unirlas hace falta una **tabla puente**: un diccionario que traduzca de un sistema al otro.

## 8.1. Una advertencia importante

Podrías pensar que los códigos numéricos de Trade Map son el estándar ISO 3166-1 numérico y que basta con una tabla ISO. **No lo son del todo.** Trade Map hereda los códigos de UN Comtrade, que introduce variantes propias:

| País | Código en Trade Map | Código ISO 3166-1 |
|---|---|---|
| Estados Unidos | `842` | `840` |
| Francia | `251` | `250` |
| India | `699` | `356` |
| Suiza | `757` | `756` |
| Noruega | `579` | `578` |

Si construyes el puente con una tabla ISO genérica y no verificas, **pierdes silenciosamente a Estados Unidos, Francia e India** del análisis. No aparece ningún error: simplemente esos países quedan sin PIB y desaparecen de las tablas finales.

Por eso construimos el puente a mano para los países que nos interesan y, sobre todo, **auditamos el resultado**.

```python
# Puente de codigos: Trade Map (numerico, estilo Comtrade) -> Banco Mundial (ISO alfa-3)
# Cubre los mayores importadores y exportadores mundiales de HS 060312.
PUENTE_CODIGOS = {
    # Mayores importadores
    "842": "USA",  # Estados Unidos  (ojo: no es 840)
    "528": "NLD",  # Paises Bajos
    "392": "JPN",  # Japon
    "616": "POL",  # Polonia
    "826": "GBR",  # Reino Unido
    "276": "DEU",  # Alemania
    "724": "ESP",  # Espana
    "410": "KOR",  # Corea del Sur
    "124": "CAN",  # Canada
    "251": "FRA",  # Francia         (ojo: no es 250)
    "643": "RUS",  # Rusia
    "764": "THA",  # Tailandia
    "246": "FIN",  # Finlandia
    "579": "NOR",  # Noruega         (ojo: no es 578)
    "784": "ARE",  # Emiratos Arabes Unidos
    "381": "ITA",  # Italia
    "757": "CHE",  # Suiza           (ojo: no es 756)
    "203": "CZE",  # Republica Checa
    "752": "SWE",  # Suecia
    "642": "ROU",  # Rumania
    "208": "DNK",  # Dinamarca
    "704": "VNM",  # Viet Nam
    "040": "AUT",  # Austria
    "036": "AUS",  # Australia
    "372": "IRL",  # Irlanda
    "620": "PRT",  # Portugal
    "056": "BEL",  # Belgica
    "804": "UKR",  # Ucrania
    # Mayores exportadores
    "170": "COL",  # Colombia
    "792": "TUR",  # Turkiye
    "156": "CHN",  # China
    "218": "ECU",  # Ecuador
    "404": "KEN",  # Kenya
    "504": "MAR",  # Marruecos
    "231": "ETH",  # Etiopia
    "320": "GTM",  # Guatemala
}

print(f"El puente cubre {len(PUENTE_CODIGOS)} paises.")
```

```python
# Aplicamos el puente: .map() traduce cada codigo usando el diccionario
comercio["iso3"] = comercio["codigo_tm"].map(PUENTE_CODIGOS)

con_puente = comercio["iso3"].notna().sum()
print(f"Paises con traduccion : {con_puente} de {len(comercio)}")
```

Nunca aceptes un cruce de tablas sin revisarlo. La pregunta que hay que hacerse siempre es: **¿quién se quedó por fuera y me importa?**

Fíjate en un caso interesante: **Taipéi Chino** (código `490`) aparece sin traducción, y no es un descuido. Se documenta y se sigue.

> **La costumbre profesional:** después de cada cruce, revisa quién quedó afuera y cuánto pesa. Reportarlo en una nota metodológica es lo que separa un análisis defendible de uno que se cae en la primera pregunta difícil.

## 9.4. Control de calidad antes de entregar

Ningún conjunto de datos sale de nuestras manos sin revisión. La instrucción `assert` es un guardián automático: si la condición que le pones no se cumple, detiene el cuaderno con un mensaje. Es la forma de que un error no pase inadvertido.

Estas siete comprobaciones son el equivalente, en datos, a la revisión final de un informe antes de enviarlo.

```python
print("CONTROL DE CALIDAD\n" + "=" * 60)

# 1. No quedo ningun agregado disfrazado de pais
assert CODIGO_MUNDO not in conjunto_1["codigo_tm"].values, "El agregado Mundo sigue en la tabla."
print("1. Sin agregados mezclados con paises .................... OK")

# 2. Un pais no aparece dos veces
assert conjunto_1["codigo_tm"].duplicated().sum() == 0, "Hay codigos de pais repetidos."
print("2. Sin paises repetidos ................................. OK")

# 3. Los valores monetarios no pueden ser negativos
assert (conjunto_1["imp_usd"] >= 0).all(), "Hay importaciones negativas."
assert (conjunto_1["exp_usd"] >= 0).all(), "Hay exportaciones negativas."
print("3. Sin valores monetarios negativos ..................... OK")

# 4. La balanza que calculamos coincide con la definicion X - M
assert np.allclose(conjunto_1["balanza_usd"],
                   conjunto_1["exp_usd"] - conjunto_1["imp_usd"]), "La balanza no cuadra."
print("4. Balanza coherente con X - M .......................... OK")

# 5. Las participaciones mundiales suman aproximadamente 100 %
suma_part = conjunto_1["imp_participacion_mundial_pct"].sum()
assert 99 <= suma_part <= 101, f"Las participaciones suman {suma_part:.2f} %."
print(f"5. Participaciones suman {suma_part:.2f} % ......................... OK")

# 6. Todos los paises del caso tienen su dato macroeconomico
assert conjunto_3["pib_usd"].notna().all(), "Falta el PIB de algun pais del caso."
print("6. Todos los paises del caso tienen PIB ................. OK")

# 7. Todos los datos macro son del mismo anio
assert conjunto_3["anio"].nunique() == 1, "Los paises del caso mezclan anios distintos."
print(f"7. Todos los datos macro son de {ANIO_REFERENCIA} ..................... OK")

print("=" * 60)
print("Los tres conjuntos estan listos.")
```

# 10. Un primer vistazo visual

Un gráfico no reemplaza el análisis, pero ayuda a ver de golpe lo que una tabla dice en varias líneas. Hagamos el más simple y más útil de todos: el ranking de compradores.

Usamos **barras horizontales** porque lo que queremos comparar son magnitudes con nombres largos, y horizontal es la única orientación en que los nombres se leen sin girar la cabeza. Marcamos en color distinto al país candidato, para que salte a la vista dónde está parado frente a los grandes.

```python
# Paleta de colores del curso (validada para lectura accesible)
AZUL      = "#2a78d6"    # serie principal
NARANJA   = "#eb6834"    # elemento destacado
GRIS_TEXT = "#52514e"
GRIS_EJE  = "#898781"
REJILLA   = "#e1e0d9"

datos = conjunto_1.nlargest(12, "imp_usd").iloc[::-1]   # invertimos para que el mayor quede arriba
colores = [NARANJA if c == CODIGO_CANDIDATO else AZUL for c in datos["codigo_tm"]]

fig, ax = plt.subplots(figsize=(9, 6))
barras = ax.barh(datos["pais_corto"], datos["imp_usd"] / 1e6, color=colores, height=0.68)

# Etiqueta directa al final de cada barra: evita que el lector tenga que ir al eje
for barra, valor in zip(barras, datos["imp_usd"] / 1e6):
    ax.text(valor + 3, barra.get_y() + barra.get_height() / 2,
            f"{valor:,.0f}", va="center", fontsize=9, color=GRIS_TEXT)

ax.set_xlabel("Importaciones en 2025 (millones de USD)", fontsize=10, color=GRIS_TEXT)
ax.set_title("Quien compra claveles frescos en el mundo\nHS 060312, doce mayores importadores",
             fontsize=13, color="#0b0b0b", loc="left", pad=14)
ax.xaxis.grid(True, color=REJILLA, linewidth=0.8)
ax.set_axisbelow(True)
ax.yaxis.grid(False)
for lado in ["top", "right", "left"]:
    ax.spines[lado].set_visible(False)
ax.spines["bottom"].set_color(GRIS_EJE)
ax.tick_params(colors=GRIS_EJE, labelsize=9)
ax.set_xlim(0, datos["imp_usd"].max() / 1e6 * 1.15)

plt.figtext(0.01, -0.02,
            "En naranja: Corea del Sur, el destino candidato.\n"
            "Fuente: elaboracion propia con datos de Trade Map (ITC, 2025).",
            fontsize=8, color=GRIS_EJE, ha="left")
plt.tight_layout()
plt.show()
```

El gráfico deja planteada la tensión del caso: Corea del Sur es un mercado claramente más pequeño que los cinco primeros. Si la decisión se tomara solo con este gráfico, la respuesta sería "no vayas a Corea, ve a Estados Unidos".

Pero la empresa ya está en Estados Unidos y precisamente quiere dejar de depender de allí. **El tamaño del mercado es solo una de las variables.** En el Cuaderno 2 vamos a agregar las otras: qué tan importador neto es cada país, qué tan abierta es su economía, a qué precio paga y a qué ritmo crece. ## Las cinco ideas para llevarse

1. **Los totales no son observaciones.** La fila "Mundo" de Trade Map y los agregados regionales del Banco Mundial son sumas, no países. Separarlos siempre.
2. **Un dato faltante no es un cero**, salvo que tengas evidencia. Nosotros la conseguimos usando la columna de balanza como prueba de consistencia.
3. **Las unidades se revisan antes de comparar.** Un precio en USD por tonelada y otro en USD por unidad no se comparan, aunque estén en la misma columna.
4. **Todo cruce se audita.** Después de unir dos tablas, la primera pregunta es siempre quién quedó afuera y cuánto pesa.
5. **Se compara en el mismo año.** Tomar el último dato disponible de cada país mezcla diferencias entre países con diferencias entre momentos.
